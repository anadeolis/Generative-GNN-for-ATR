```python
import csv
import ast
import networkx as nx
from itertools import combinations
from tqdm import tqdm

def exact_atr(G, p=0.5):
    edges = list(G.edges())
    m = len(edges)
    total_prob = 0.0

    for k in range(m + 1):
        for subset in combinations(edges, k):
            H = nx.Graph()
            H.add_nodes_from(G.nodes())
            H.add_edges_from(subset)
            if nx.is_connected(H):
                prob = (p**k) * ((1 - p)**(m - k))
                total_prob += prob
    return total_prob

def run_atr_across_all_p(input_file='graphs_n8.csv', output_file='exact_atr_surface.csv'):
    p_values = [round(i * 0.01, 2) for i in range(1, 100)]  # p from 0.01 to 0.99

    with open(input_file, 'r') as f_in, open(output_file, 'w', newline='') as f_out:
        reader = csv.DictReader(f_in)
        writer = csv.writer(f_out)
        writer.writerow(['num_edges', 'p', 'atr', 'edge_list'])

        for row in tqdm(reader, desc="Reading graphs"):
            edge_list = ast.literal_eval(row['edge_list'])
            G = nx.Graph()
            G.add_nodes_from(range(8))  # assuming 8 nodes
            G.add_edges_from(edge_list)
            num_edges = len(edge_list)

            for p in p_values:
                atr = exact_atr(G, p)
                writer.writerow([num_edges, p, atr, edge_list])
                f_out.flush()

    print("Done: All ATRs written to", output_file)

run_atr_across_all_p()
```
